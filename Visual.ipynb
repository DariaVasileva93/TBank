```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
%matplotlib inline
```

```python
df=pd.read_csv('C:\\code\\WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# TotalCharges to float
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
#coerce - все пропуски заменятся на NaN. Nan имеет тип float
```

```python
df.info()
```

```python
plt.hist(df['MonthlyCharges'])
plt.show()
```

```python
plt.hist(df['MonthlyCharges'], edgecolor='k')
plt.title('Распределение месячной платы за услуги')
plt.xlabel('плата в месяц, долл.')
plt.ylabel('кол-во наблюдений')
plt.grid()
```

```python
pic = sns.histplot(data=df, x='MonthlyCharges')
pic.set(xlabel="плата в месяц, долл.", ylabel="кол-во наблюдений", title="Распределение месячной платы за услуги")
plt.grid()
```

```python
pic = sns.histplot(data=df, x='MonthlyCharges', hue='SeniorCitizen')
pic.set(xlabel="плата в месяц, долл.", ylabel="кол-во наблюдений", title="Распределение месячной платы за услуги")
plt.grid()
```

```python
# создание объекта - гистограммы
# nbinsx - параметр, которым можно задать кол-во бакетов, на котрые хотим разбить распределение

hist= go.Histogram(x=df['MonthlyCharges'], nbinsx=20)

# создание объекта Figure, в него передается гистограмма
fig = go.Figure(data=hist)

fig.update_layout(
    width=600, # размер графика
    height=400,
    title='Распределение месячной платы за услуги',
    xaxis_title='плата в месяц, долл.',
    yaxis_title='кол-во наблюдений'
)
fig.show()
```

```python
# как зависят в среднем суммарные выплаты клиента от времени жизни с компанией
# считаем, сколько в среднем нам заплатит 1 клиент в зависимости от кол-ва месяцев использования наших услуг
df1=df.groupby('tenure').agg({'TotalCharges':'mean'}).reset_index()
```

```python
df1.head()
```

```python
line = go.Scatter(x=df1['tenure'], y=df1['TotalCharges'], mode='lines')
fig = go.Figure(data=line)

fig.update_layout(
    autosize=False,
    width=700,
    height=400,
    title='Зависимость средних выплат клиента от времени жизни с компанией',
    xaxis_title='месяцев с подкл.услугами',
    yaxis_title='средние выплаты клиента'
)
fig.show()
```
